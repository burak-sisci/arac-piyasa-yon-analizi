Bu notebook, `down < stable < up` kümülatif ayrıştırmasının test-dışı rolling-origin denetimini gösterir. Kilitli test kapalıdır.

## Monoton olasılık sözleşmesi

Ham `q1=P(y>down)` ve `q2=P(y>stable)` için `q1<q2` ise ikisi `(q1+q2)/2` değerine projekte edilir. Sonra `p_down=1-q1`, `p_stable=q1-q2`, `p_up=q2` hesaplanır.

```python
import json
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
MODEL_DIR = ROOT / 'data' / 'processed' / 'model'
sonuc = json.loads((MODEL_DIR / 'model_15_frank_hall_ordinal_ozet.json').read_text(encoding='utf-8'))
sonuc['karar'], sonuc['test'], sonuc['ortam'], sonuc['ordinal_denetim']
```

```python
pd.DataFrame(sonuc['genel_metrikler']).T.sort_values('mcc', ascending=False)
```
